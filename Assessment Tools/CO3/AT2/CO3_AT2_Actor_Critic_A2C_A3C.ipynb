## 1. Mathematical Formulation & Objectives
Actor-Critic methods optimize policy $\pi_\theta(a|s)$ and value $V_\phi(s)$ simultaneously:

$$\mathcal{L}_{actor} = -\log \pi_\theta(a|s) A(s,a) - \beta H(\pi_\theta), \quad \mathcal{L}_{critic} = \frac{1}{2} (R + \gamma V(s') - V(s))^2$$
